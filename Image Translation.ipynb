```python
# Import 
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

# **Image Translation**



```python
# Read the image using OpenCV
img = cv2.imread("baseball.jpg")
```

```python
# Display the image using Matplotlib
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis("off")
```

```python
# Translation 
M = np.float32([[1, 0, 50],
                [0, 1, 80]])
translated = cv2.warpAffine(img, M, (636,438))

# Display original and translated images
plt.figure(figsize=(10,5))

plt.subplot(1,2,1)
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,2,2)
plt.imshow(translated[:,:,::-1])
plt.title("Translated Image")
plt.axis("off")

plt.show()
```

# **Image Scaling**



```python
# Scale the image
small_img = cv2.resize(img, None, fx=5, fy=2)
```

```python
large_img = cv2.resize(img, None, fx=10, fy=50)
```

```python
# Display images
plt.figure(figsize=(18,6))

plt.subplot(1,3,1)
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(small_img[:,:,::-1])
plt.title("Downscaled Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(large_img[:,:,::-1])
plt.title("Upscaled Image")
plt.axis("off")

plt.show()
```

```python
print("Original Shape :", img.shape)
print("Downscaled Shape :", small_img.shape)
print("Upscaled Shape :", large_img.shape)
```

# **Image Shearing**



```python
# Get image height and width
h, w = img.shape[:2]
```

```python
# Horizontal shearing matrix
M1 = np.float32([[1, 0.5, 0],
                 [0, 1, 0]])

# Vertical shearing matrix
M2 = np.float32([[1, 0, 0],
                 [0.5, 1, 0]])
```

```python
# Apply shearing
h_shear = cv2.warpAffine(img, M1, (w, h))
v_shear = cv2.warpAffine(img, M2, (w, h))
```

```python
# Display images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(h_shear[:,:,::-1])
plt.title("Horizontal Shear")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(v_shear[:,:,::-1])
plt.title("Vertical Shear")
plt.axis("off")

plt.show()
```

# **Image Reflection**



```python
# Horizontal reflection
h_flip = cv2.flip(img, 1)

# Vertical reflection
v_flip = cv2.flip(img, 0)

# Both-axis reflection
hv_flip = cv2.flip(img, -1)
```

```python
# Display images
plt.figure(figsize=(10,10))

plt.subplot(2,2,1)
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis("off")

plt.subplot(2,2,2)
plt.imshow(h_flip[:,:,::-1])
plt.title("Horizontal Reflection")
plt.axis("off")

plt.subplot(2,2,3)
plt.imshow(v_flip[:,:,::-1])
plt.title("Vertical Reflection")
plt.axis("off")

plt.subplot(2,2,4)
plt.imshow(hv_flip[:,:,::-1])
plt.title("Both-axis Reflection")
plt.axis("off")

plt.show()
```

# **Image Rotation**



```python
# Get image height and width
h, w = img.shape[:2]
```

```python
# Find center of image
center = (w//2, h//2)
```

```python
# Rotation matrix for 45 degree
M1 = cv2.getRotationMatrix2D(center, 45, 1)
```

```python
# Rotation matrix for 90 degree
M2 = cv2.getRotationMatrix2D(center, 90, 1)
```

```python
# Rotate images
rotated_45 = cv2.warpAffine(img, M1, (w, h))
rotated_90 = cv2.warpAffine(img, M2, (w, h))
```

```python
# Display images
plt.figure(figsize=(15,5))

plt.subplot(1,3,1)
plt.imshow(img[:,:,::-1])
plt.title("Original Image")
plt.axis("off")

plt.subplot(1,3,2)
plt.imshow(rotated_45[:,:,::-1])
plt.title("45° Rotated Image")
plt.axis("off")

plt.subplot(1,3,3)
plt.imshow(rotated_90[:,:,::-1])
plt.title("90° Rotated Image")
plt.axis("off")

plt.show()
```
